```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

model_id = "Qwen/Qwen3-0.6B"

tokenizer = AutoTokenizer.from_pretrained(model_id)

model = AutoModelForCausalLM.from_pretrained(
    model_id,
    torch_dtype="auto",
    device_map="auto"
)

messages = [
    {"role": "user", "content": "Explain transformers in simple words."}
]

inputs = tokenizer.apply_chat_template(
    messages,
    add_generation_prompt=True,
    tokenize=True,
    return_dict=True,
    return_tensors="pt"
).to(model.device)

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=200,
        do_sample=True,
        temperature=0.6,
        top_p=0.95,
        top_k=20
    )

response = tokenizer.decode(
    outputs[0][inputs["input_ids"].shape[-1]:],
    skip_special_tokens=True
)

print(response)
```

## Level 3

```python
import os
import json
import torch
from safetensors.torch import load_file
from transformers import PreTrainedTokenizerFast
from transformers.models.qwen3.configuration_qwen3 import Qwen3Config
from transformers.models.qwen3.modeling_qwen3 import Qwen3ForCausalLM

local_dir = "./Qwen3-0.6B-local"

def token_value(x):
    if isinstance(x, dict):
        return x.get("content")
    return x

with open(os.path.join(local_dir, "config.json"), "r") as f:
    config_dict = json.load(f)

config = Qwen3Config(**config_dict)

with open(os.path.join(local_dir, "tokenizer_config.json"), "r") as f:
    tokenizer_config = json.load(f)

tokenizer_kwargs = {}

for key in ["bos_token", "eos_token", "unk_token", "pad_token"]:
    value = tokenizer_config.get(key)
    if value is not None:
        tokenizer_kwargs[key] = token_value(value)

additional_special_tokens = tokenizer_config.get("additional_special_tokens")

if additional_special_tokens is not None:
    tokenizer_kwargs["additional_special_tokens"] = [
        token_value(x) for x in additional_special_tokens
    ]

tokenizer = PreTrainedTokenizerFast(
    tokenizer_file=os.path.join(local_dir, "tokenizer.json"),
    **tokenizer_kwargs
)

tokenizer.chat_template = tokenizer_config.get("chat_template")

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

model = Qwen3ForCausalLM(config)

state_dict = load_file(
    os.path.join(local_dir, "model.safetensors"),
    device="cpu"
)

model.load_state_dict(state_dict, strict=True)

device = "cuda" if torch.cuda.is_available() else "cpu"
dtype = torch.float16 if device == "cuda" else torch.float32

model = model.to(device=device, dtype=dtype)
model.eval()

messages = [
    {"role": "user", "content": "Explain transformers in simple words."}
]

inputs = tokenizer.apply_chat_template(
    messages,
    add_generation_prompt=True,
    tokenize=True,
    return_dict=True,
    return_tensors="pt"
).to(device)

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=200,
        do_sample=True,
        temperature=0.6,
        top_p=0.95,
        top_k=20
    )

response = tokenizer.decode(
    outputs[0][inputs["input_ids"].shape[-1]:],
    skip_special_tokens=True
)

print(response)
```

## Level 4

```python
import os
import json
import torch
import regex as re
from safetensors.torch import load_file
from transformers.models.qwen3.configuration_qwen3 import Qwen3Config
from transformers.models.qwen3.modeling_qwen3 import Qwen3ForCausalLM

local_dir = "./Qwen3-0.6B-local"

def bytes_to_unicode():
    bs = (
        list(range(ord("!"), ord("~") + 1))
        + list(range(ord("¡"), ord("¬") + 1))
        + list(range(ord("®"), ord("ÿ") + 1))
    )
    cs = bs[:]
    n = 0

    for b in range(256):
        if b not in bs:
            bs.append(b)
            cs.append(256 + n)
            n += 1

    return dict(zip(bs, map(chr, cs)))

def get_pairs(word):
    pairs = set()
    prev_char = word[0]

    for char in word[1:]:
        pairs.add((prev_char, char))
        prev_char = char

    return pairs

def token_value(x):
    if isinstance(x, dict):
        return x.get("content")
    return x

with open(os.path.join(local_dir, "config.json"), "r") as f:
    config_dict = json.load(f)

with open(os.path.join(local_dir, "tokenizer_config.json"), "r") as f:
    tokenizer_config = json.load(f)

with open(os.path.join(local_dir, "vocab.json"), "r", encoding="utf-8") as f:
    encoder = json.load(f)

with open(os.path.join(local_dir, "merges.txt"), "r", encoding="utf-8") as f:
    merges = f.read().splitlines()

merges = [line for line in merges if line and not line.startswith("#")]
merges = [tuple(line.split()) for line in merges]
bpe_ranks = dict(zip(merges, range(len(merges))))

decoder = {v: k for k, v in encoder.items()}

added_token_to_id = {}
added_id_to_token = {}
special_token_ids = set()

for token_id_str, item in tokenizer_config["added_tokens_decoder"].items():
    token_id = int(token_id_str)
    content = item["content"]

    added_token_to_id[content] = token_id
    added_id_to_token[token_id] = content

    encoder[content] = token_id
    decoder[token_id] = content

    if item["special"]:
        special_token_ids.add(token_id)

all_added_tokens = sorted(
    added_token_to_id.keys(),
    key=len,
    reverse=True
)

byte_encoder = bytes_to_unicode()
byte_decoder = {v: k for k, v in byte_encoder.items()}

pat = re.compile(
    r"""(?i:'s|'t|'re|'ve|'m|'ll|'d)|[^\r\n\p{L}\p{N}]?\p{L}+|\p{N}| ?[^\s\p{L}\p{N}]+[\r\n]*|\s*[\r\n]+|\s+(?!\S)|\s+"""
)

bpe_cache = {}

def bpe(token):
    if token in bpe_cache:
        return bpe_cache[token]

    word = tuple(token)

    if len(word) == 1:
        return token

    pairs = get_pairs(word)

    while True:
        bigram = min(
            pairs,
            key=lambda pair: bpe_ranks.get(pair, float("inf"))
        )

        if bigram not in bpe_ranks:
            break

        first, second = bigram
        new_word = []
        i = 0

        while i < len(word):
            try:
                j = word.index(first, i)
                new_word.extend(word[i:j])
                i = j
            except ValueError:
                new_word.extend(word[i:])
                break

            if (
                i < len(word) - 1
                and word[i] == first
                and word[i + 1] == second
            ):
                new_word.append(first + second)
                i += 2
            else:
                new_word.append(word[i])
                i += 1

        word = tuple(new_word)

        if len(word) == 1:
            break

        pairs = get_pairs(word)

    result = " ".join(word)
    bpe_cache[token] = result
    return result

def split_by_added_tokens(text):
    if not all_added_tokens:
        return [text]

    pattern = "(" + "|".join(re.escape(tok) for tok in all_added_tokens) + ")"
    return [part for part in re.split(pattern, text) if part != ""]

def encode_normal_text(text):
    ids = []

    for token in re.findall(pat, text):
        token = "".join(byte_encoder[b] for b in token.encode("utf-8"))
        bpe_tokens = bpe(token).split(" ")

        for bpe_token in bpe_tokens:
            ids.append(encoder[bpe_token])

    return ids

def encode(text):
    ids = []

    for part in split_by_added_tokens(text):
        if part in added_token_to_id:
            ids.append(added_token_to_id[part])
        else:
            ids.extend(encode_normal_text(part))

    return ids

def decode(ids, skip_special_tokens=True):
    output = []
    byte_stream = bytearray()

    def flush_bytes():
        nonlocal byte_stream
        if byte_stream:
            output.append(byte_stream.decode("utf-8", errors="replace"))
            byte_stream = bytearray()

    for idx in ids:
        idx = int(idx)

        if skip_special_tokens and idx in special_token_ids:
            continue

        token = decoder[idx]

        if idx in added_id_to_token:
            flush_bytes()
            output.append(token)
            continue

        for char in token:
            if char in byte_decoder:
                byte_stream.append(byte_decoder[char])
            else:
                flush_bytes()
                output.append(char)

    flush_bytes()
    return "".join(output)

def build_chat_prompt(messages, add_generation_prompt=True, enable_thinking=True):
    prompt = ""

    if len(messages) > 0 and messages[0]["role"] == "system":
        prompt += (
            "<|im_start|>system\n"
            + messages[0]["content"]
            + "<|im_end|>\n"
        )
        start_index = 1
    else:
        start_index = 0

    for message in messages[start_index:]:
        role = message["role"]
        content = message["content"]

        prompt += (
            "<|im_start|>"
            + role
            + "\n"
            + content
            + "<|im_end|>\n"
        )

    if add_generation_prompt:
        prompt += "<|im_start|>assistant\n"

        if enable_thinking is False:
            prompt += "<think>\n\n</think>\n\n"

    return prompt

config = Qwen3Config(**config_dict)

model = Qwen3ForCausalLM(config)

state_dict = load_file(
    os.path.join(local_dir, "model.safetensors"),
    device="cpu"
)

model.load_state_dict(state_dict, strict=True)

device = "cuda" if torch.cuda.is_available() else "cpu"
dtype = torch.float16 if device == "cuda" else torch.float32

model = model.to(device=device, dtype=dtype)
model.eval()

messages = [
    {"role": "user", "content": "Explain transformers in simple words."}
]

prompt = build_chat_prompt(
    messages,
    add_generation_prompt=True,
    enable_thinking=True
)

input_ids_list = encode(prompt)

inputs = {
    "input_ids": torch.tensor(
        [input_ids_list],
        dtype=torch.long
    ).to(device),
    "attention_mask": torch.ones(
        (1, len(input_ids_list)),
        dtype=torch.long
    ).to(device)
}

eos_token_id = added_token_to_id["<|im_end|>"]
pad_token_id = added_token_to_id["<|endoftext|>"]

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=200,
        do_sample=True,
        temperature=0.6,
        top_p=0.95,
        top_k=20,
        eos_token_id=eos_token_id,
        pad_token_id=pad_token_id
    )

generated_ids = outputs[0][inputs["input_ids"].shape[-1]:].tolist()

response = decode(
    generated_ids,
    skip_special_tokens=True
)

print(response)
```

```python
import torch
from transformers import AutoProcessor, AutoModelForImageTextToText

model_id = "Qwen/Qwen3-VL-2B-Instruct"
image_path = "sample_car.jpg"

processor = AutoProcessor.from_pretrained(model_id)

model = AutoModelForImageTextToText.from_pretrained(
    model_id,
    dtype="auto",
    device_map="auto"
)

messages = [
    {
        "role": "user",
        "content": [
            {"type": "image", "image": image_path},
            {"type": "text", "text": "Describe this image in detail."}
        ]
    }
]

inputs = processor.apply_chat_template(
    messages,
    tokenize=True,
    add_generation_prompt=True,
    return_dict=True,
    return_tensors="pt"
).to(model.device)

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=128
    )

generated_ids = outputs[0][inputs["input_ids"].shape[-1]:]

response = processor.decode(
    generated_ids,
    skip_special_tokens=True,
    clean_up_tokenization_spaces=False
)

print(response)
```

## Level1

```python
import os
import json
import glob
import torch
from huggingface_hub import snapshot_download
from safetensors.torch import load_file

from transformers import Qwen2TokenizerFast
from transformers.models.qwen3_vl.configuration_qwen3_vl import Qwen3VLConfig
from transformers.models.qwen3_vl.modeling_qwen3_vl import Qwen3VLForConditionalGeneration
from transformers.models.qwen3_vl.processing_qwen3_vl import Qwen3VLProcessor
from transformers.models.qwen3_vl.video_processing_qwen3_vl import Qwen3VLVideoProcessor
from transformers.models.qwen2_vl.image_processing_qwen2_vl import Qwen2VLImageProcessor


model_id = "Qwen/Qwen3-VL-2B-Instruct"
local_dir = "./Qwen3-VL-2B-Instruct-local"
image_path = "sample_car.jpg"


if not os.path.exists(local_dir):
    snapshot_download(
        repo_id=model_id,
        local_dir=local_dir
    )


def token_value(x):
    if isinstance(x, dict):
        return x.get("content")
    return x


with open(os.path.join(local_dir, "config.json"), "r") as f:
    config_dict = json.load(f)

with open(os.path.join(local_dir, "tokenizer_config.json"), "r") as f:
    tokenizer_config = json.load(f)

with open(os.path.join(local_dir, "preprocessor_config.json"), "r") as f:
    processor_config = json.load(f)


config = Qwen3VLConfig(**config_dict)


tokenizer_kwargs = {}

for key in ["bos_token", "eos_token", "unk_token", "pad_token"]:
    value = tokenizer_config.get(key)
    if value is not None:
        tokenizer_kwargs[key] = token_value(value)

additional_special_tokens = tokenizer_config.get("additional_special_tokens")

if additional_special_tokens is not None:
    tokenizer_kwargs["additional_special_tokens"] = [
        token_value(x) for x in additional_special_tokens
    ]


tokenizer = Qwen2TokenizerFast(
    vocab_file=os.path.join(local_dir, "vocab.json"),
    merges_file=os.path.join(local_dir, "merges.txt"),
    tokenizer_file=os.path.join(local_dir, "tokenizer.json"),
    **tokenizer_kwargs
)

tokenizer.chat_template = tokenizer_config.get("chat_template")

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token


image_processor_config = dict(processor_config)
image_processor_config.pop("image_processor_type", None)
image_processor_config.pop("video_processor_type", None)
image_processor_config.pop("processor_class", None)

video_processor_config = dict(processor_config)
video_processor_config.pop("image_processor_type", None)
video_processor_config.pop("video_processor_type", None)
video_processor_config.pop("processor_class", None)


image_processor = Qwen2VLImageProcessor(**image_processor_config)
video_processor = Qwen3VLVideoProcessor(**video_processor_config)


processor = Qwen3VLProcessor(
    image_processor=image_processor,
    tokenizer=tokenizer,
    video_processor=video_processor,
    chat_template=tokenizer.chat_template
)


model = Qwen3VLForConditionalGeneration(config)


safetensor_files = sorted(glob.glob(os.path.join(local_dir, "*.safetensors")))

state_dict = {}

for path in safetensor_files:
    shard = load_file(path, device="cpu")
    state_dict.update(shard)

missing, unexpected = model.load_state_dict(state_dict, strict=False)

missing = [k for k in missing if k != "lm_head.weight"]

if missing:
    print("Missing keys:", missing)

if unexpected:
    print("Unexpected keys:", unexpected)


device = "cuda" if torch.cuda.is_available() else "cpu"

if device == "cuda":
    dtype = torch.bfloat16 if torch.cuda.is_bf16_supported() else torch.float16
else:
    dtype = torch.float32

model = model.to(device=device, dtype=dtype)
model.eval()


pad_token_id = tokenizer.convert_tokens_to_ids("<|endoftext|>")
eos_token_id = tokenizer.convert_tokens_to_ids("<|im_end|>")

model.generation_config.pad_token_id = pad_token_id
model.generation_config.eos_token_id = eos_token_id


messages = [
    {
        "role": "user",
        "content": [
            {"type": "image", "image": image_path},
            {"type": "text", "text": "Describe this image in detail."}
        ]
    }
]


inputs = processor.apply_chat_template(
    messages,
    tokenize=True,
    add_generation_prompt=True,
    return_dict=True,
    return_tensors="pt"
)


inputs = {
    k: v.to(device) if torch.is_tensor(v) else v
    for k, v in inputs.items()
}


with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=128,
        pad_token_id=pad_token_id,
        eos_token_id=eos_token_id
    )


generated_ids = outputs[0][inputs["input_ids"].shape[-1]:]

response = processor.decode(
    generated_ids,
    skip_special_tokens=True,
    clean_up_tokenization_spaces=False
)

print(response)
```
